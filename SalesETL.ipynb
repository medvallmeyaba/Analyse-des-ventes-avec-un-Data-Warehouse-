# load data into DataWarhouse

```python
df3["order_date"] = pd.to_datetime(df3["order_date"])

date = df3[["order_date"]].drop_duplicates()

date["date_key"] = date["order_date"].dt.strftime("%Y%m%d").astype(int)
date["full_date"] = date["order_date"]
date["year"] = date["order_date"].dt.year
date["month"] = date["order_date"].dt.month
date["day"] = date["order_date"].dt.day

date = date[
    ["date_key", "full_date", "year", "month", "day"]
]

```

```python
df1.to_sql(
    "dim_customer",
    engine,
    if_exists="append",
    index=False
)

df2.to_sql(
    "dim_product",
    engine,
    if_exists="append",
    index=False
)

date.to_sql(
    "dim_date",
    engine,
    if_exists="append",
    index=False
)
```

```python
# Get keys form tables in database
dim_customer = pd.read_sql(
    "SELECT customer_key, customer_id FROM dim_customer",
    engine
)

dim_product = pd.read_sql(
    "SELECT product_key, product_id FROM dim_product",
    engine
)

dim_date=pd.read_sql(
    "SELECT full_date , day FROM dim_date",
    engine
)
```

```python
# Join 3 table into fact

fact = df3.merge(
    dim_customer,
    on="customer_id",
    how='left'
).merge(
    dim_product,
    on="product_id",
    how='left'
)

fact
```

```python
fact = fact[
    [
        "order_id",
        "customer_key",
        "product_key",
        "date_key",
        "quantity",
        "total_amount"
    ]
] 
```

```python
fact.to_sql(
     "fact_orders",
     engine,
     if_exists="append",
     index=False
)
```
